**Preprocessing**

```python
Features = ['dt','dur','tot_dur','pktrate','port_no','rx_kbps','tot_kbps']
X = df[Features]
```

```python
kolom_negatif = ['dt','dur','tot_dur']
X = X[(X[kolom_negatif]>=0).all(axis=1)]
```

```python
df = df.loc[X.index]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
num_features = ['dt','dur','tot_dur','pktrate','rx_kbps','tot_kbps']
num_transform = scaler
```

```python
import numpy as np

np.random.seed(42)
sample_indices = np.random.choice(X.shape[0], size=1000000, replace=False)
X_sample = X.iloc[sample_indices]
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(transformers=[
    ('num', num_transform, num_features)
], remainder = 'passthrough')
```

```python
X_scaled = preprocessor.fit_transform(X_sample)
```

```python
type(X_scaled)
```

```python
X_scaled = X_scaled.astype(np.float32)
input_dim = X_scaled.shape[1]
```

**Training**

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
import keras_tuner as kt

def build_autoencoder(hp):
  model = Sequential()
  model.add(Input(shape=[input_dim]))
  hp_activation = hp.Choice('activation', values=['relu', 'elu', 'tanh'])

  # Tuning Jumlah Layer pada Encoder
  num_encoder_layers = hp.Int('num_encoder_layers', min_value = 1, max_value=4)
  encoder_units = []

  # Membangun layer encoder secara dinamis sesuai jumlah yang terpilih
  for i in range(num_encoder_layers):
    units = hp.Int(f'enc_units_{i}', min_value=32, max_value=128, step=16)
    model.add(Dense(units=units, activation=hp_activation))
    encoder_units.append(units)

  # Tuning Layer Bottleneck / Kompresi Terdalam
  hp_bottleneck = hp.Int('bottleneck', min_value= 4, max_value=64, step=4)
  model.add(Dense(units=hp_bottleneck, activation=hp_activation))

  # Membangun Layer Decoder (Harus sama persis bentuk tapi urutannya dibalik)
  for i in reversed(range(num_encoder_layers)):
    model.add(Dense(units=encoder_units[i], activation=hp_activation))

  # Output Layer
  model.add(Dense(input_dim, activation='linear'))

  # Tuning Learning Rate
  hp_learning_rate = hp.Choice('learning_rate', values = [1e-2, 1e-3, 1e-4])

  model.compile(optimizer=Adam(learning_rate=hp_learning_rate),loss = 'mse')
  return model
```

```python
tuner = kt.Hyperband(
    build_autoencoder,
    objective='val_loss',
    max_epochs=30,
    factor=3,
    directory='autoencoder_tuning',
    project_name='iot_anomaly'
)
```

```python
callbacks = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)
```

```python
tuner.search(
    X_scaled, X_scaled,
    epochs=30,
    validation_split = 0.1,
    batch_size=1024,
    callbacks = [callbacks],
    verbose = 1
)
```

```python
best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]
```

```python
best_autoencoder = tuner.hypermodel.build(best_hps)
best_autoencoder.summary()
```

```python
callbacks_retraining = [ EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True
)
]
```

```python
history = best_autoencoder.fit(
    X_scaled, X_scaled,
    epochs=50,
    validation_split=0.2,
    batch_size=1024,
    callbacks=callbacks_retraining,
    verbose=1
)
```

```python
best_tuning = tuner.get_best_models(num_models=1)[0]
```

```python
X_pred1 = best_tuning.predict(X_scaled, batch_size=1024)
```

```python
X_pred = best_autoencoder.predict(X_scaled, batch_size=1024)
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

mse = np.mean(np.power(X_scaled - X_pred, 2), axis=1)
mse1 = np.mean(np.power(X_scaled - X_pred1, 2), axis=1)
```

```python
threshold = np.percentile(mse, 99)
threshold1= np.percentile(mse1, 99)
print(f'Nilai Threshold (99th percentile): {threshold:.4f}')
print(f'Nilai Threshold 1 (99th percentile): {threshold1:.4f}')
```

```python
label_ae = np.where(mse > threshold, 1,0)
label_ae1 = np.where(mse1 > threshold1, 1,0)
df_tuning = pd.DataFrame(X_scaled)
df_build = pd.DataFrame(X_scaled)
```

```python
df_tuning['label'] = label_ae1
df_build['label'] = label_ae
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=3, random_state=42)
X_pca = pca.fit_transform(X_eval1)
```

```python
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2','PC3'])
df_pca['cluster'] = label_eval1
df_pca['cluster'] = df_pca['cluster'].astype(str)
```

```python
import seaborn as sns

warna_pallete = {'0':'blue', '1':'red'}

fig_1 = plt.figure(figsize=(20,8))
ax1 = fig_1.add_subplot(121)
sns.scatterplot(
    x = 'PC1', y='PC2',
    hue='cluster',
    data=df_pca,
    palette=warna_pallete,
    alpha=0.6,
    s=40,
    ax=ax1
)
ax1.set_title('AutoEncoder: Distribusi 2D', fontsize=12, fontweight='bold')
ax1.set_xlabel('Principal Component 1')
ax1.set_ylabel('Principal Component 2')
ax1.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
fig_2 = plt.figure(figsize=(20,8))
ax2 = fig_2.add_subplot(122, projection='3d')
for cluster_id in df_pca['cluster'].unique():
  subset = df_pca[df_pca['cluster'] == cluster_id]

  # Pengaturan ukuran dan urutan tumpukan agar anomali lebih terlihat
  if cluster_id == '1':
    ukuran = 25
    z_order= 5
  else:
    ukuran = 15
    z_order = 1

  ax2.scatter(
      subset['PC1'], subset['PC2'], subset['PC3'],
      label = f'Cluster {cluster_id}',
      c=warna_pallete.get(cluster_id, 'gray'),
      alpha= 0.6,
      s=ukuran,
      zorder=z_order
  )
ax2.set_title('AutoEncoder: Distribusi 3D', fontsize=12, fontweight='bold')
ax2.set_xlabel('PC 1')
ax2.set_ylabel('PC 2')
ax2.set_zlabel('PC 3')
ax2.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
