**Preprocessing**

```python
Features = ['dt', 'dur', 'tot_dur', 'pktrate', 'port_no', 'rx_kbps', 'tot_kbps']
X = df[Features]
```

```python
kolom_negatif = ['dt', 'dur', 'tot_dur']
X = X[(X[kolom_negatif] >=0).all(axis=1)]
```

```python
print(X.shape)
print(df.shape)
```

```python
df = df.loc[X.index]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
import numpy as np

np.random.seed(42)
sample_indices = np.random.choice(X_scaled.shape[0], size=20000, replace=False)
X_sample = X_scaled[sample_indices]
```

```python
from sklearn.neighbors import NearestNeighbors

k = 20 # Min_samples
neighbors = NearestNeighbors(n_neighbors=k, n_jobs=-1)
neighbors.fit(X_sample)
```

```python
distances, indices = neighbors.kneighbors(X_sample)
```

```python
k_distances = np.sort(distances[:,k-1], axis=0)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.plot(k_distances, color='blue', linewidth=2)
plt.title('K-Distance Graph', fontsize=14)
plt.xlabel('Titik Data (Diurutkan berdasarkan jarak ke tetangga-14)', fontsize=12)
plt.ylabel('Jarak (eps)', fontsize=12)
plt.ylim(0, np.percentile(k_distances, 99))
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

**Training**

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score

# 1. Rentang eksplorasi parameter
eps_uji = [0.5, 0.8, 1.0, 1.5, 2.0, 3.0]
min_samples_uji = [20, 30, 50, 100]

skor_terbaik = -1
parameter_juara = {}

print("Mencari parameter DBSCAN paling optimal secara objektif...\n")

for e in eps_uji:
    for ms in min_samples_uji:
        model_uji = DBSCAN(eps=e, min_samples=ms, n_jobs=-1)
        label_uji = model_uji.fit_predict(X_sample)

        # 2. Filter noise untuk evaluasi objektif antar-cluster
        mask_uji = label_uji != -1
        label_bersih = label_uji[mask_uji]
        X_bersih = X_sample[mask_uji]

        # Evaluasi HANYA bisa dilakukan jika terbentuk minimal 2 cluster valid (di luar noise)
        jumlah_cluster_valid = len(set(label_bersih))

        if jumlah_cluster_valid > 1:
            sil = silhouette_score(X_bersih, label_bersih)
            db = davies_bouldin_score(X_bersih, label_bersih)

            jumlah_noise = list(label_uji).count(-1)

            print(f"eps={e:.2f} | ms={ms} -> {jumlah_cluster_valid} Cluster | Noise: {jumlah_noise} | Sil: {sil:.4f} | DB: {db:.4f}")

            # 3. Wasit Penentu: Silhouette Score tertinggi adalah juaranya
            if sil > skor_terbaik:
                skor_terbaik = sil
                parameter_juara = {
                    'eps': e,
                    'min_samples': ms,
                    'jml_cluster': jumlah_cluster_valid,
                    'jml_noise': jumlah_noise,
                    'skor_sil': sil,
                    'skor_db': db
                }

# 4. Penobatan Juara
print("\n" + "="*50)
print("🏆 PARAMETER JUARA DITEMUKAN 🏆")
print("="*50)
if parameter_juara:
    print(f"eps Terbaik         : {parameter_juara['eps']}")
    print(f"min_samples Terbaik : {parameter_juara['min_samples']}")
    print(f"Membentuk           : {parameter_juara['jml_cluster']} Cluster")
    print(f"Jumlah Noise        : {parameter_juara['jml_noise']} Titik")
    print(f"Silhouette Score    : {parameter_juara['skor_sil']:.4f}")
    print(f"Davies-Bouldin      : {parameter_juara['skor_db']:.4f}")
else:
    print("Tidak ditemukan kombinasi yang menghasilkan lebih dari 1 cluster.")
```

```python
from sklearn.cluster import DBSCAN

eps = 1.50
dbscan = DBSCAN(eps=eps, min_samples=k, n_jobs=-1)
cluster_labels = dbscan.fit_predict(X_sample)
```

```python
jumlah_cluster = len(set(cluster_labels)) - (1 if -1 in cluster_labels else 0)
jumlah_noise = list(cluster_labels).count(-1)
persentase_noise = (jumlah_noise / len(cluster_labels)) * 100
print(f'Jumlah Cluster: {jumlah_cluster}')
print(f'Jumlah Noise : {jumlah_noise} dan persentase noise {persentase_noise}')
```

**Evaluate**

```python
df_profil = pd.DataFrame(X_sample, columns=Features)
df_profil['cluster'] = cluster_labels
```

```python
ukuran_cluster = df_profil.groupby('cluster').size().reset_index(name='Jumlah Paket')
rata_rata_fitur = df_profil.groupby('cluster').mean().reset_index()
tabel_analisis = pd.merge(ukuran_cluster, rata_rata_fitur, on='cluster')
tabel_analisis = tabel_analisis.sort_values(by='Jumlah Paket', ascending=False)
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

mask = cluster_labels !=-1
X_cluster_only = X_sample[mask]
labels_cluster_only = cluster_labels[mask]
```

```python
sil_score = silhouette_score(X_cluster_only, labels_cluster_only)
db_score = davies_bouldin_score(X_cluster_only, labels_cluster_only)
ch_score = calinski_harabasz_score(X_cluster_only, labels_cluster_only)
print(f"Silhouette Score: {sil_score:.4f}")
print(f"Davies-Bouldin Score: {db_score:.4f}")
print(f"Calinski-harabasz Score: {ch_score:.4f}")
```

```python
from sklearn.decomposition import PCA
import seaborn as sns

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_sample)
```

```python
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2', 'PC3'])
df_pca['cluster'] = cluster_labels
df_pca['cluster'] = df_pca['cluster'].astype(str)
```

```python
fig_1 = plt.figure(figsize=(20,8))
ax1 = fig_1.add_subplot(121)
sns.scatterplot(
    x='PC1', y='PC2',
    hue='cluster',
    data=df_pca,
    palette='Set1',
    alpha= 0.7,
    s=40,
    ax=ax1
)
ax1.set_title('Distribusi Cluster 2D (PCA1 vs PCA2)', fontsize=15, fontweight='bold')
ax1.set_xlabel('Principal Component 1')
ax1.set_ylabel('Principal Component 2')
ax1.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
fig_2 = plt.figure(figsize=(25,10))
ax2 = fig_2.add_subplot(111, projection='3d')
for cluster_id in df_pca['cluster'].unique():
  subset = df_pca[df_pca['cluster'] == cluster_id]

  if cluster_id == '-1':
    warna = 'black'
    label_nama = 'Noise (-1)'
    ukuran = 25
    z_order = 5
  else:
    warna = None
    label_nama = f'Cluster {cluster_id}'
    ukuran = 15
    z_order = 1
  ax2.scatter(
      subset['PC1'],subset['PC2'], subset['PC3'],
      label = label_nama,
      c=warna,
      alpha = 0.6,
      s=ukuran,
      zorder=z_order
  )
ax2.set_title('Distribusi Cluster 3D', fontsize=15, fontweight='bold')
ax2.set_xlabel('PC 1')
ax2.set_ylabel('PC 2')
ax2.set_zlabel('PC 3')
ax2.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
explained_variance = pca.explained_variance_ratio_.sum()*100
print(f"\nTotal informasi (variasi) dari 7 fitur asli yang berhasil dipertahankan oleh 3 PC: {explained_variance:.2f}")
```
